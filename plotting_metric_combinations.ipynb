```python

def construct_df(model, show=False):
    scores_path= path_scores(model)+"mrr@20__Instruct-Query-template.json"
    with open(scores_path) as f:
        scores = json.load(f)
    with open(path(model)+"prompt_geometry.json") as f:
        data1 = json.load(f)
    if show:
        print(data1.keys())
    df_scores = pd.DataFrame.from_dict(scores, orient="index", columns=["score"])
    #columns= [f"prompt{i}" for i in range(len(scores.values()))])
    df_scores = df_scores.reset_index().rename(columns={"index": "prompt_text"})
    df_angle = pd.DataFrame.from_dict(data1).T
    df = df_scores.merge(df_angle, on='prompt_text')
    prompt_dict = prompts(dataset)
    df["prompt_label"] = df["prompt_text"].apply(lambda prompt: prompt_dict[prompt])
    if show:
        display(df.head())
        display(df.tail())
    return df

#construct_df(models[0], show=True)
```

```python

def plot(df, x, y="score", colors="prompt_label", sizes="score", title="", legend_title=None):

    # legend title that explains formatting
    if legend_title is None:
        legend_title = f"colors:{colors}, size:{sizes}"

    # Normalize scores for marker size
    min_size, max_size = 10, 30
    try:
        # parse the value from dictionary
        df["sizes"] = df[sizes].apply(lambda d: float(d['mean']))
        ranks = df["sizes"].rank(method='average')
    except:   # for non-dict format: i.e. prompt_label or score
        ranks = df[sizes].rank(method='average')
    marker_sizes = (
        (ranks - ranks.min()) / (ranks.max() - ranks.min()) * (max_size - min_size) + min_size
    )

    if y == "score":
        # non-dict format
        y_vals = df[y].apply(lambda d: float(d))
        y_err = [0 for i in y_vals]
    else:
        y_vals = df[y].apply(lambda d: float(d['mean']))
        y_err = df[y].apply(lambda d: float(d['std']))

    x_vals = df[x].apply(lambda d: float(d['mean']))
    x_err  = df[x].apply(lambda d: float(d['std']))


    # Create figure
    fig = go.Figure()
    fig.add_trace(
        go.Scatter(
            mode='markers',
            x=x_vals,
            y=y_vals,
            error_x=dict(type='data', array=x_err, visible=True, color='lightgray'),  # optional std bars
            error_y=dict(type='data', array=y_err, visible=True, color='lightgray'),
            marker=dict(
                size=marker_sizes,
            #    colorscale='Cividis',
                color=df[colors] if df[colors].dtype==float else df[colors].apply(lambda d: float(d['mean'])),
                colorbar=dict(title=f"C:{colors} S:{sizes}"),
                showscale=True,
            ),
            text=df['prompt_text'],
            hovertemplate=(
                '<b>Prompt:</b> %{text}<br>'
                '<b>X (distance):</b> %{x:.4f}<br>'
                '<b>Y (score):</b> %{y:.4f}<br>'
            ),
        ),
    )


    # Add one trace per alpha value

    fig.update_layout(
        title = title,
        xaxis_title=x,#'Cos-distance compared to Q-A line',
        yaxis_title=y,#'Prompt performance',
        height=600,
        width=1000,
        template="none",
    )
    fig.update_layout(legend_title_text=legend_title)


    fig.show()
```

What each metric tells you
| Metric	| Question it answers |
|--------|--------|
|chord_similarity	|Does the prompt push q in the same direction as a?|
|sim_improvement	|Does the prompt make pq closer to a? (the practical question)|
|displacement	|How much does the prompt change the embedding?|
|parallel_magnitude	|How much movement is toward the answer? (signed: negative = away)|
|orthogonal_magnitude	|How much movement is sideways (perpendicular to the q→a axis)?|
|parallel_fraction	|What fraction of total movement goes toward the answer?|

The decomposition into parallel and orthogonal components is probably the most directly useful for your question. If you find that parallel_fraction is low (close to 0) even when sim_improvement is positive, that tells you the relationship is indeed complex — the prompt isn't pushing straight toward the answer but arriving at a better position through a largely orthogonal path.
